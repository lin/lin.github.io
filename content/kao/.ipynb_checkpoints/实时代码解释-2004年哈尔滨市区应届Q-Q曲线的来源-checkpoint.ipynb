# 哈尔滨市区应届生的点是怎么来的？

页面里的蓝点回答：**2004 年哈尔滨市区的中考尖子，三年后的高考比大庆的预测高多少或低多少？**
哈尔滨和大庆的中考用**不同的试卷**（哈尔滨满分 590，大庆满分 690），分数不能直接比。所以只能比**位置**：

1. 哈尔滨市区中考的**第 N 个位置**，对应大庆市区中考的**第 N × 21500 ÷ 37811 名**（按考生人数比例折算）；
2. 用上一个 notebook 里的红线，由大庆市区位次得到「预测高考分」；
3. 取已核实的哈尔滨市区应届生，把他们的高考分从高到低排，**第 N 高的高考分**就是「实际」。
4. 实际 − 预测 = 这个位置上哈尔滨比大庆曲线高或低多少。

这是一个 Q-Q 图：两列各自排好序，只对比同名次，**不对应具体的人**。

## 0 读数据

```python
import math
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
from pathlib import Path

plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'Hiragino Sans GB', 'Heiti TC', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

def find(name):   # 在 content/kao 里运行，或从仓库根目录运行均可
    return next(p for p in [Path(name), Path('content/kao')/name] if p.exists())
rd = lambda name: pd.read_csv(find(name), encoding='utf-8-sig')

# ---- 红线：大庆市区应届 Q-Q 曲线（做法见《大庆市区应届 Q-Q 曲线的来源》）----
roster = rd('2004年大庆中考录取总表.csv')
roster['文体'] = pd.to_numeric(roster['文化课+体育成绩'])
city_rank = lambda z: int((roster['文体'] > z).sum() + 1)          # 中考分 → 大庆市区位次
ys = rd('2007年高考理科一分一段表.csv').sort_values('累计人数')
score_of_m = lambda m: float(np.interp(m, ys['累计人数'], ys['分数']))   # 全省位次 → 高考分
ANCH = [(2, 16.4), (12, 84), (30, 225), (56, 339), (87, 510), (129, 693), (177, 942), (234, 1227), (274, 1598)]

def m_of_n(n):
    # 市区应届位次 n → 全省位次 m：锚点间 log-log 线性；n < 2 时沿第一段延长；n > 274 返回 None
    if n > ANCH[-1][0]: return None
    i = 0
    while i < len(ANCH) - 2 and n > ANCH[i + 1][0]: i += 1
    (n0, m0), (n1, m1) = ANCH[i], ANCH[i + 1]
    w = (math.log(n) - math.log(n0)) / (math.log(n1) - math.log(n0))
    return math.exp(math.log(m0) + w * (math.log(m1) - math.log(m0)))

g_of_n = lambda n: None if m_of_n(n) is None else score_of_m(m_of_n(n))      # 位次 → 预测高考分

# 红线按中考分查表（623–660 逐分），之后对任意小数分线性插值
tab = pd.DataFrame({'z': range(623, 664)})
tab['g'] = [g_of_n(city_rank(z)) for z in tab['z']]
red = lambda z: float(np.interp(z, tab['z'], tab['g'])) if 623 <= z <= 663 else np.nan

front100 = rd('2007年高考哈尔滨市前100名.csv')       # 2007 哈尔滨高考前 100 名（理科）
perp = rd('2004年哈尔滨中考换算大庆等效中考（逐人，60组）.csv')   # 能配上 2004 哈尔滨中考名次的 60 人
page = rd('2004年哈尔滨中考换算大庆等效中考（Q-Q位置，67个）.csv')   # 页面使用的 Q-Q 位置表：用来核对
page.columns = ['N', 'r', 'z', 'p_old', 'pc', 'g', 'd', 'inc']
N_HRB, N_DQ = 37811, 21500                           # 哈尔滨市区、大庆市区中考考生数（页面给出）
print(len(front100), '人在哈尔滨高考前 100 名；其中应届', int((front100['应往届'] == '应届').sum()), '人')
```

## 1 先要决定：用哪些人？

高考前 100 名里有 95 位应届生，但**并不是每个人都参加了 2004 年哈尔滨市区中考**（有的来自县里、外地，有的是转学来的）。
只有「能确认参加了 2004 年哈尔滨市区中考」的应届生，才有资格代表这个总体——否则我们比的就不是同一批人。
页面确认了 **67 位**：名录里配对到中考名次的 60 人，名录里重名未能配上的吴琼、刘聪、杨旭、于歌，以及六九联中升入哈三中的任彧、王平平、赵瑜珩。

```python
extra = ['吴琼', '刘聪', '杨旭', '于歌', '任彧', '王平平', '赵瑜珩']
names = list(perp['姓名']) + extra
verified = front100[front100['姓名'].isin(names)]
g_sorted = np.sort(verified['总分'].to_numpy())[::-1]       # 第 N 高的高考分
assert len(verified) == 67 and (g_sorted == page['g'].to_numpy()).all()
print(f'已核实 {len(verified)} 人；高考分从高到低排序后，与页面「第 N 高」逐位一致 ✔')
print('前 12 个位置的高考分：', g_sorted[:12].tolist())
print('高考前 100 名里未列入核实名单的应届生：', int((front100['应往届'] == '应届').sum()) - len(verified), '人（不画；页面说明里提到其中来历不明的有 10 人）')
```

## 2 位置对位置：哈尔滨第 N 个 → 大庆第 r 名

$$ r = N \times \frac{21500}{37811} \approx 0.569\,N $$

哈尔滨市区考生 37811 人，大庆市区 21500 人，所以哈尔滨每 1 个位置对应大庆 0.569 名。
**这一步假设哈尔滨市区考生和大庆市区考生的「能力分布」相同**——这是个假设，不是事实，后面会说。

```python
N = np.arange(1, 68)
r = N * N_DQ / N_HRB
print('最大偏差（与页面「换算大庆市区位次」）：', np.abs(r.round(2) - page['r'].to_numpy()).max().round(3))
pred = np.array([g_of_n(x) for x in r])                      # 大庆红线给出的预测高考分
print('与页面「校正曲线预测高考」最大相差：', np.abs(pred.round(1) - page['pc'].to_numpy()).max().round(2), '分（出现在最前几名，第 1–2 个位置）')
out = pd.DataFrame({'哈市位置 N': N, '换算大庆位次 r': r.round(2), '预测高考': pred.round(1), '实际(第N高)': g_sorted,
                    '实际−预测': (g_sorted - pred).round(1)})
out.iloc[[0, 1, 2, 4, 9, 19, 29, 39, 49, 66]].set_index('哈市位置 N').T
```

## 3 大庆等效中考分（只是帮助看图）

把「大庆市区第 r 名」翻译回大庆的中考分数——在大庆市区位次表里反查（并列处取插值）。它只用来决定散点画在横轴的哪个位置，**不影响预测高考分**（预测直接由 r 得到）。

```python
zz = np.arange(600, 664)
n_tab = np.array([city_rank(z) for z in zz])
def z_of_r(r_):                                         # 位次 → 中考分（反查位次表）
    o = np.argsort(n_tab, kind='stable')
    return float(np.interp(r_, n_tab[o], zz[o]))
z_eq = np.array([z_of_r(x) for x in r])
print('与页面「大庆等效中考」：', (np.abs(z_eq - page['z'].to_numpy()) < .15).mean().round(2) * 100, '% 的位置相差 < 0.15 分，最大相差', np.abs(z_eq - page['z'].to_numpy()).max().round(1), '分（并列平台处）')
```

## 4 出图

```python
use = N >= 2                                    # 第 1 位（718 分）是异常值：不画、不计入统计
diff = g_sorted - pred
fig, ax = plt.subplots(1, 3, figsize=(16, 4.8))
BLUE, RED, GREY, SKY = '#1f6fb2', '#c0392b', '#7f8c8d', '#00a8ff'

# ① Q-Q 点 + 红线
a = ax[0]
zr = np.linspace(623, 660, 200); a.plot(zr, [red(z) for z in zr], color=RED, lw=2, label='大庆市区应届曲线（红线）')
a.scatter(z_eq[use], g_sorted[use], s=14, color=SKY, zorder=3, label='哈尔滨市区应届生（66 个位置）')
a.scatter(z_eq[:1], g_sorted[:1], s=40, facecolor='none', edgecolor=GREY, label='第 1 位 718 分（异常，不计）')
a.set_xlabel('大庆等效中考分'); a.set_ylabel('高考分'); a.set_title('① 哈尔滨的 Q-Q 点落在红线附近'); a.legend(fontsize=8); a.grid(alpha=.3)

# ② 每个位置的差
a = ax[1]
a.bar(N[use], diff[use], color=np.where(diff[use] >= 0, BLUE, RED), alpha=.8)
a.axhline(diff[use].mean(), color='k', ls='--', lw=1, label=f'平均 {diff[use].mean():+.1f} 分')
a.set_xlabel('哈尔滨中考位置 N'); a.set_ylabel('实际 − 预测（分）'); a.set_title('② 逐位置的差：整体略高，靠前的位置偏低'); a.legend(fontsize=8); a.grid(alpha=.3)

# ③ 对比：同一批人「逐人」比 vs 「位置」比
a = ax[2]
pp = perp.copy()
for c_ in ['大庆等效中考(满分690)', '校正曲线预测高考', '高考−预测']: pp[c_] = pd.to_numeric(pp[c_], errors='coerce')
pp = pp.dropna(subset=['校正曲线预测高考'])
a.scatter(pp['大庆等效中考(满分690)'], pp['高考−预测'], s=18, color='#e67e22', label=f'逐人（{len(pp)} 人）：平均 {pp["高考−预测"].mean():+.1f}')
a.scatter(z_eq[use], diff[use], s=14, color=SKY, label=f'按位置（66 个）：平均 {diff[use].mean():+.1f}')
a.axhline(0, color='k', lw=.8)
a.set_xlabel('大庆等效中考分'); a.set_ylabel('实际 − 预测（分）'); a.set_title('③ 逐人比高约 15 分，按位置比只高约 2 分'); a.legend(fontsize=8); a.grid(alpha=.3)
fig.suptitle('哈尔滨市区应届生 Q-Q 点是怎么来的', fontsize=14)
fig.tight_layout(); fig.savefig('2004年哈尔滨市区应届Q-Q曲线的来源.png', dpi=150); plt.show()
```

## 5 读图：「逐人比」与「按位置比」为什么差这么多？

- **按位置（蓝点）**：把两边各自排好序，比的是「同名次的人群」——哈尔滨第 N 个位置 vs 大庆第 r 名。平均只比曲线高约 1.6 分，基本持平。
- **逐人（橙点）**：每个人的哈尔滨中考名次折算成大庆位次，预测他的高考，再与他**自己的**高考分相减。平均高出约 15 分。

橙点高的原因是**选择偏差**：这 60 个人是从「高考哈尔滨前 100 名」里挑出来的——他们**高考考得特别好**，所以高考减预测几乎一定是正的
（这类样本不能当作「同一批中考生的整体表现」）。Q-Q 的做法不看具体的人，只比分布，所以不会被这种选择放大。

## 6 结论与边界

1. **哈尔滨点的做法**：按人数比例把哈尔滨中考位置折成大庆位次（r = N × 21500 ÷ 37811），再用大庆红线预测，与已核实应届生的高考分第 N 高相比。
2. **能说**：把哈尔滨第 N 个位置对到大庆第 r 名，哈尔滨市区应届尖子的高考表现与大庆曲线基本持平（平均 +1.6 分，靠前几个位置略低、其余略高）。
3. **不能说**：
   - **等比例换算是假设**：它要求哈尔滨市区考生和大庆市区考生的能力分布相同。如果哈尔滨的分布更宽、顶端更厚，换算就会低估哈尔滨的位置。这个假设无法由这里的数据检验；
   - **名单有选择性**：只收了高考进哈尔滨前 100 名（≥678）的应届生，越靠后的位置越不完整、越偏高；
   - **37811 与 21500 是页面给出的总人数**，本 notebook 没有独立核对来源；
   - **第 1 位 718 分**按页面做法当异常值剔除，如果计入，平均会略高。
